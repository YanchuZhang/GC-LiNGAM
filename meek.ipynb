```python
conf_str = [
    [(1,2)],                                 # 1 edge
    [(1,2),(1,3)],                          # 2 edges
    [(1,2),(1,3),(1,4)],                    # 3 edges: star
    [(1,2),(1,3),(2,3)],                    # 3 edges: triangle + isolated
    [(1,2),(1,3),(2,4)],                    # 3 edges: path P4
    [(1,2),(1,3),(2,3),(1,4)],             # 4 edges: triangle + pendant
    [(1,2),(1,3),(1,4),(2,3)],             # 4 edges: 4-cycle with diagonal
    [(1,2),(1,3),(1,4),(2,3),(2,4)],      # 5 edges: K4 minus one edge
]
```

```python
len(conf_str)
```

```python
def all_dags(n):
    vertices = range(1, n + 1)

    edges = [(i, j) for i in vertices for j in vertices if i != j]

    dags = []

    for mask in range(1 << len(edges)):
        G = nx.DiGraph()
        G.add_nodes_from(vertices)

        for k, edge in enumerate(edges):
            if mask & (1 << k):
                G.add_edge(*edge)

        if nx.is_directed_acyclic_graph(G):
            dags.append(set(G.edges()))

    return dags

dags = all_dags(4)
dags
```

```python
confounding_count = 1
stop = False

for B in conf_str:

    if stop:
        break

    # print(f"Processing Confounding Structure: {confounding_count}/{len(conf_str)}")

    dags_copy = dags

    while len(dags_copy) > 0:
        G_temp = dags_copy[0]
        res_meek = find_equivalence_class_meek(G_temp,B,n)
        res_exh = find_equivalent_graphs_exhaustive_to_compare(G_temp,B,n)
        if frozenset(map(frozenset, res_meek)) != frozenset(map(frozenset, res_exh)):
            print("!!!BAD CASE!!!")
            print(res_meek)
            print(res_exh)
            stop = True
            break
        else:
            # print(res_meek)
            # print(res_exh)
            dags_copy = [x for x in dags_copy if x not in res_meek]
            print(f"Remaining DAG: {len(dags_copy)}/{len(dags)}, Confounding id: {confounding_count}/{len(conf_str)}")

        confounding_count = confounding_count + 1
```
