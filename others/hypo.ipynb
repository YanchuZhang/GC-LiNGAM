```python
from sympy import symbols, Matrix, eye
```

```python
n = 8

'''
l = {(i, j): symbols(f"l{i}{j}")
     for i in range(1, n + 1)
     for j in range(i + 1, n + 1)}
'''

b = {(i, j): symbols(f"b{i}{j}")
     for i in range(1, n + 1)
     for j in range(1, n + 1)
     if i != j}

m = {(i, j): symbols(f"m{i}{j}")
     for i in range(1, n + 1)
     for j in range(1, n + 1)
     if i != j}

'''
L = Matrix([
    [
        l[(i, j)] if i < j else 0
        for j in range(1, n + 1)
    ]
    for i in range(1, n + 1)
])
'''

M = Matrix([
    [
        m[(i, j)] if i < j else 0
        for j in range(1, n + 1)
    ]
    for i in range(1, n + 1)
])

B = Matrix([
    [
        b[(i, j)] if i > j else 0
        for j in range(1, n + 1)
    ]
    for i in range(1, n + 1)
])

for i in range(n):
    B[(i,i)]=1

u = 3
v = 6

M[(v-1,u-1)] = m[(v,u)]
M[(u-1,v-1)] = 0

B, M
```

```python
I = eye(n)
A = (I-M).T * B
A
```
